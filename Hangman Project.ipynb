```python
from IPython.display import clear_output
import random

# Function that holds the main menu loop of the game and passes information to other menu functions, including word banks and the leaderboard
def start_menu():
    # word bank to extract a word from when hangman game difficulty is set to normal
    normal_word_bank = [
        'apple',
        'beach',
        'bread',
        'chair',
        'cloud',
        'dance',
        'dream',
        'eagle',
        'flame',
        'floor',
        'fruit',
        'grass',
        'green',
        'house',
        'light',
        'money',
        'music',
        'ocean',
        'paper',
        'plant',
        'queen',
        'river',
        'round',
        'shirt',
        'smile',
        'snake',
        'space',
        'stone',
        'table',
        'train'
    ]

    # word bank to extract a word from when hangman game difficulty is set to hard
    hard_word_bank = [
        'awkward',
        'bizarre',
        'blizzard',
        'cryptic',
        'dazzling',
        'exhaust',
        'frenzy',
        'giraffe',
        'jigsaw',
        'jukebox',
        'knuckle',
        'mystery',
        'oxygen',
        'quartz',
        'rhythm',
        'sphinx',
        'squawk',
        'strange',
        'subway',
        'symbol',
        'thicket',
        'vortex',
        'whiskey',
        'wretched',
        'zephyr',
        'zigzag',
        'zombie',
        'journey',
        'jackpot',
        'complex'
    ]

    # dictionary holding player statistics to be displayed as a leaderboard
    leaderboard = {
        'Fabio': {
            'normal': {
                'games_played': 4,
                'wins': 3,
                'best_remaining_guesses': 4
            },
            'hard':{
                'games_played': 2,
                'wins': 0,
                'best_remaining_guesses': 0 
            }
        },
        'Alexandra': {
            'normal': {
                'games_played': 15,
                'wins': 14,
                'best_remaining_guesses': 6
            },
            'hard':{
                'games_played': 12,
                'wins': 9,
                'best_remaining_guesses': 4 
            }
        },
        'Bertrand': {
            'normal': {
                'games_played': 5,
                'wins': 1,
                'best_remaining_guesses': 2
            },
            'hard':{
                'games_played': 12,
                'wins': 2,
                'best_remaining_guesses': 3 
            }
        }
    }

    continue_playing = True
    hard_difficulty = False
    player_name = ''

    # continue running the menu while the player has not chosen to quit the game
    while continue_playing:
        # if the player does not have a name, call the get_player_name function so they can choose one
        if not player_name:
            player_name = get_player_name()
            
        # display all the required menu information
        menu_info(player_name, hard_difficulty)
        
        # retrieves the player's chosen action from the menu
        continue_playing, hard_difficulty, select_leaderboard, play_game = get_menu_input(continue_playing, hard_difficulty)
        
        if play_game:
            # if hard difficulty is selected play hangman using a hard word and update the leaderboard
            if hard_difficulty:
                current_guesses, win = play_hangman(hard_word_bank[random.randint(0,29)])
                leaderboard = update_leaderboard(leaderboard, player_name, hard_difficulty, current_guesses, win)
            else:
                current_guesses, win = play_hangman(normal_word_bank[random.randint(0,29)])
                leaderboard = update_leaderboard(leaderboard, player_name, hard_difficulty, current_guesses, win)
        elif select_leaderboard:
            show_leaderboard(leaderboard)

        # clears the printed output avoid overwhelming the player with information
        clear_output()  
            

# Function that retrieves the name of the current player that started the menu
# Returns:
# player_name - string containing the name chosen by the player
def get_player_name():
    while True:
        # gets the players chosen name
        player_name = input('Enter your name (maximum of 15 characters):').strip()
        clear_output()   

        # Checks that player_name is not long or empty
        if len(player_name) > 15:
            print('Please keep your name below 15 characters')
        elif not player_name:
            Print('Please enter a name')
        else:
            return player_name


# Function that prints information related to the main menu and the actions that can be taken on it
# Parameters:
# player_name - string containing the name chosen by the player
# hard_difficulty - boolean where True/False indicates whether the player wants to play the hard/normal version of hangman
def menu_info(player_name, hard_difficulty):
    print(f'Hi {player_name}, welcome to Hangman, this is a game where you must guess a mystery word one letter at a time.')
    if hard_difficulty:
        print('You are on Hard difficulty')
    else:
        print('You are on Normal difficulty')
    print('You can swap between Normal and Hard difficulties by typing "normal" or "hard" in the current menu')
    print('You can view statistics of past games and previous players by typing "leaderboard"')
    print('You can play Hangman by typing "play"')
    print('And you can quit the game by typing "quit"')


# Function that returns the player's action on the main menu
# Parameters:
# continue_playing - boolean where True/False indicates whether the player wants to continue/stop playing
# hard_difficulty - boolean where True/False indicates whether the player wants to play the hard/normal version of hangman
# Returns:
# continue_playing - boolean where True/False indicates whether the player wants to continue/stop playing
# hard_difficulty - boolean where True/False indicates whether the player wants to play the hard/normal version of hangman
# select_leaderboard - boolean where True indicates whether the player wants to view the leaderboard
# play_game - boolean where True indicates whether the player wants to play a game of hangman
def get_menu_input(continue_playing, hard_difficulty):
    while True:
        # gets the players chosen menu command
        action = input('What do you want to do?').strip()   

        # selects the requested menu option
        if action == 'normal':
            hard_difficulty = False
            return continue_playing, hard_difficulty, False, False 
        elif action == 'hard':
            hard_difficulty = True
            return continue_playing, hard_difficulty, False, False 
        elif action == 'leaderboard':
            return continue_playing, hard_difficulty, True, False 
        elif action == 'play':
            return continue_playing, hard_difficulty, False, True 
        elif action == 'quit':
            continue_playing = False
            return continue_playing, hard_difficulty, False, False 
        else:
            print('Please enter one of the following commands: play, normal, hard, leaderboard or quit')


# Function that holds the main loop of a single hangman game and calls the relevant functions throughout the game
# Parameters:
# word - string containing the hangman word to be guessed
# Returns:
# current_guesses - integer indicating the number of extra guesses the player ended the game with
# win - boolean where True/False indicates whether the player won/lost the hangman game they played
def play_hangman(word):
    # initialises required variables
    guessed_letters = []
    guessed_word = list('-' * len(word))
    total_guesses = 6
    current_guesses = 6
    correct_guesses = 0

    # game continues until the player is out of guesses or they have correctly guessed the word
    while current_guesses > 0 and correct_guesses != len(word):
        # clears the printed output every time a round has passed
        clear_output()
        # prints information related to the game
        game_info(current_guesses, total_guesses, guessed_letters, guessed_word)
        # asks the player to guess a letter
        letter = get_guess(guessed_letters)
        # checks whether the letter matches a letter in the hangman word
        current_guesses, correct_guesses, guessed_word, guessed_letters = process_guess(letter, word, current_guesses, correct_guesses, guessed_word, guessed_letters)    
    # return game statistics so they can be inserted into the leaderboard
    return check_win(correct_guesses, word, current_guesses)    
    

# Function that prints information about remaining guesses, guessed letter and the guessed word on the current round of hangman
# Parameters:
# current_guesses - integer indicating the number of extra guesses the player ended the game with
# total_guesses - integer indicating the total number of guesses the player has at the start of the game
# guessed_letters - list containing the letters the player has currently guessed
# guessed_word - list representing individual letters of the hangman word with filled in indexes where the player guessed the letters correctly
def game_info(current_guesses, total_guesses, guessed_letters, guessed_word):
    # List that stores ASCII for current hangman round
    hangman_stages = [
        '''
         +---+
         |   |
         O   |
        /|\\  |
        / \\  |
             |
        =========
        ''',
        '''
         +---+
         |   |
         O   |
        /|\\  |
        /    |
             |
        =========
        ''',
        '''
         +---+
         |   |
         O   |
        /|\\  |
             |
             |
        =========
        ''',
        '''
         +---+
         |   |
         O   |
        /|   |
             |
             |
        =========
        ''',
        '''
         +---+
         |   |
         O   |
             |
             |
             |
        =========
        ''',
        '''
         +---+
         |   |
             |
             |
             |
             |
        =========
        '''
    ]
    if current_guesses == total_guesses:
        print('You are playing Hangman! In this game you must guess the mystery word one letter at a time.')
    guessed_letters.sort()
    print(hangman_stages[current_guesses-1])
    print(f'Your current guessed word is {''.join(guessed_word)} \n You have {current_guesses} incorrect guesses before you lose! \n You have already guessed: {' '.join(guessed_letters)} ')


# Function that checks if the guess was correct or incorrect 
# Parameters:
# guessed_letters - list containing the letters the player has currently guessed
# Returns:
# letter - string containing the letter picked by the player
def get_guess(guessed_letters):
    alphabet = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
     'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
    
    while True: 
        letter = input('Enter a letter:').strip().lower()  

        # checks that the input is not a part of the alphabet or is not a single letter
        if letter not in alphabet or len(letter) != 1:
            print('Please enter a single letter')
        # if the letter was already guessed inform the player
        elif letter in guessed_letters:
            print('You have already guessed that letter')
        else:
            return letter


# Function that checks if the guessed letter was correct or incorrect 
# Parameters:
# letter - string containing the letter picked by the player
# word - string containing the hangman word to be guessed
# current_guesses - integer indicating the number of extra guesses the player ended the game with 
# correct_guesses - integer indicating the number of times the player correctly guessed a letter from the hangman word
# guessed_word - list representing individual letters of the hangman word with filled in indexes where the player guessed the letters correctly
# guessed_letters - list containing the letters the player has currently guessed
# Returns:
# current_guesses - integer indicating the updated number of extra guesses the player ended the game with 
# correct_guesses - integer indicating the updated number of times the player correctly guessed a letter from the hangman word
# guessed_word - list representing updated individual letters of the hangman word with filled in indexes where the player guessed the letters correctly
# guessed_letters - list containing the updated letters the player has currently guessed
def process_guess(letter, word, current_guesses, correct_guesses, guessed_word, guessed_letters):
    if letter in word:
        print('You guessed correctly!')
        # add the number of times the letter appears within the word to correct_guesses
        correct_guesses += word.count(letter)
        # fill in the blanks within guessed_word to show the players progress
        for i in range(len(word)):
            if word[i] == letter:
                guessed_word[i] = letter
    else:
        print('You guessed incorrectly!')
        current_guesses -= 1
    guessed_letters.append(letter)

    return current_guesses, correct_guesses, guessed_word, guessed_letters


# Function that checks if the player won or lost the hangman game and outputs their results
# Parameters:
# correct_guesses - integer indicating the number of times the player correctly guessed a letter from the hangman word
# word - string containing the hangman word to be guessed
# current_guesses - integer indicating the number of extra guesses the player ended the game with 
# Returns:
# current_guesses - integer indicating the number of extra guesses the player ended the game with
# win - boolean where True/False indicates whether the player won/lost the hangman game they played
def check_win(correct_guesses, word, current_guesses):
    # Player wins if they correctly guessed each letter of the word
    win = correct_guesses == len(word)
    if win:
        print(f'Congratulations!!! You guessed {word} with {current_guesses} guesses remaining')
    else:
        print(f'You lost. \nThe word you were guessing was {word}')
    
    # Pauses the program so the player can view their results
    pause = input('Press enter to return to the main menu')
    return current_guesses, win

    
# Function that adds player information to the leaderboard dictionary based on the previous hangman game they played
# Parameters:
# leaderboard - dictionary containing player information, including games played, wins, best remaining guesses
# player_name - string containing the name chosen by the player
# hard_difficulty - boolean where True/False indicates whether the previous hangman game the player played was on hard/normal difficulty
# current_guesses - integer indicating the number of remaining guesses the player had when the player finished the prvious hangman game
# win - boolean where True/False indicates whether the player won/lost the previous hangman game they played
# Returns:
# leaderboard - dictionary containing updated player information, including games played, wins, best remaining guesses
def update_leaderboard(leaderboard, player_name, hard_difficulty, current_guesses, win):
    # Instantiates player details if they are not present
    if player_name not in leaderboard:
        leaderboard[player_name] = {
            'normal': {
                'games_played': 0,
                'wins': 0,
                'best_remaining_guesses': 0
            },
            'hard':{
                'games_played': 0,
                'wins': 0,
                'best_remaining_guesses': 0 
            }
        }

    # game_type is dynamic to change the correct difficulty information for the player
    game_type = ''
    if hard_difficulty:
        game_type = 'hard'
    else:
        game_type = 'normal'

    # adds 1 to number of games_played
    leaderboard[player_name][game_type]['games_played'] += 1
    # adds 1 to wins if the player won their previous hangman game
    if win:
        leaderboard[player_name][game_type]['wins'] += 1
    # changes best_remaining_guesses if the player performed better in their previous hangman game
    if current_guesses > leaderboard[player_name][game_type]['best_remaining_guesses']:
        leaderboard[player_name][game_type]['best_remaining_guesses'] = current_guesses
    return leaderboard

    
# Function that prints sorted player details for the hard and normal versions of hangman
# Parameter:
# leaderboard - dictionary containing player information, including games played, wins, best remaining guesses
def show_leaderboard(leaderboard):
    # Sorts the players within the leaderboard dictionary into a tuple based on their best remaining guesses in the hard difficulty
    hard_sorted_players = sorted(leaderboard.items(), key=lambda player: player[1]['hard']['best_remaining_guesses'], reverse = True)
    # Sorts the players within the leaderboard dictionary into a tuple based on their best remaining guesses in the normal difficulty
    normal_sorted_players = sorted(leaderboard.items(), key=lambda player: player[1]['normal']['best_remaining_guesses'], reverse = True)

    # Prints out the leaderboard statistics of the hard version of hangman
    print('\nHard Leaderboard\n')
    print('Player Name  ,  Games Played  ,  Wins  ,  Best Guesses Remaining')
    for player, stats in hard_sorted_players:
        print(f' {player} , {stats['hard']['games_played']} , {stats['hard']['wins']} , {stats['hard']['best_remaining_guesses']}')

    # Prints out the leaderboard statistics of the normal version of hangman
    print('\nNormal Leaderboard\n')
    print('Player Name  ,  Games Played  ,  Wins  ,  Best Guesses Remaining')
    for player, stats in normal_sorted_players:
        print(f' {player} , {stats['normal']['games_played']} , {stats['normal']['wins']} , {stats['normal']['best_remaining_guesses']}')
    
    # Pauses the program so the player can view the leaderboard
    pause = input('Press enter to return to the main menu')
    clear_output()
    
start_menu()
```
